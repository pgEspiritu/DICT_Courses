### Step 5: Generate the course dataset
Run this exactly as-is — it creates an identical `service_requests.csv` for everyone, so no file upload or download is needed.


```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 500

barangays = ['Poblacion', 'San Isidro', 'Santa Cruz', 'Malinis', 'Bagong Silang']
service_types = ['Business Permit', 'Barangay Clearance', 'Indigency Certificate',
                  'Health Consultation', 'Social Welfare Assistance', 'Road Repair Report']
status_options = ['Pending', 'In Progress', 'Completed', 'Cancelled']
sexes = ['Male', 'Female']

data = {
    'request_id': [f'REQ-{1000+i}' for i in range(n)],
    'date_filed': pd.date_range('2025-01-01', periods=n, freq='6h').strftime('%Y-%m-%d'),
    'barangay': np.random.choice(barangays, n),
    'service_type': np.random.choice(service_types, n),
    'requestor_name': [f'Resident {i}' for i in range(n)],   # placeholder PII
    'age': np.random.randint(18, 80, n),
    'sex': np.random.choice(sexes, n),
    'contact_number': [f'09{np.random.randint(100000000, 999999999)}' for _ in range(n)],  # PII
    'priority_level': np.random.choice(['Low', 'Medium', 'High'], n, p=[0.5, 0.35, 0.15]),
    'status': np.random.choice(status_options, n, p=[0.2, 0.25, 0.5, 0.05]),
    'resolution_time_days': np.random.exponential(scale=4, size=n).round(1),
    'satisfaction_rating': np.random.choice([1, 2, 3, 4, 5], n, p=[0.05, 0.1, 0.2, 0.35, 0.3])
}

df = pd.DataFrame(data)
df.to_csv('/content/service_requests.csv', index=False)
df.head()
```

### Apply anonymization
Option A removes a direct identifier entirely. Option B masks a value, keeping only enough to verify a record if truly necessary.


```python
df_anon = df.copy()

# Option A: Remove direct identifiers entirely
df_anon = df_anon.drop(columns=['requestor_name'])

# Option B: Mask the contact number, keeping only the last 4 digits
df_anon['contact_number'] = df_anon['contact_number'].apply(
    lambda x: 'XXXXXXX' + str(x)[-4:]
)

df_anon.head()
```

### Optional stretch task: Pseudonymization
Instead of deleting the name, replace it with a code and keep the real names in a **separate, restricted** lookup table that would never be shared alongside the analysis dataset.


```python
# Restricted lookup table — keep this OFF any shared/public file in a real workflow
lookup = df[['request_id', 'requestor_name']].copy()

# Replace names in the working copy with a pseudonym code
df_anon['requestor_code'] = 'RESIDENT-' + df_anon.index.astype(str)
df_anon.head()
```

### Data Handling Note — ANSWER KEY
**Columns identified as personal/sensitive:**
`requestor_name` (direct identifier — a person's name) and `contact_number` (direct identifier — a personal phone number).
**Action taken:**
Removed `requestor_name` entirely; not needed for aggregate analysis. Masked `contact_number`, retaining only the last 4 digits for internal verification purposes. Also demonstrated pseudonymizing `requestor_name` into a `requestor_code`, with the real names kept in a separate `lookup` table that would never be distributed alongside the analysis dataset.
**Rationale:**
These fields directly identify individuals and are not required to compute service-level statistics (e.g., average resolution time per barangay). Removing/masking them reduces re-identification risk while preserving analytical value, consistent with RA 10173's principle of proportionality — only process what is necessary for the stated purpose.
**Residual risk:**
Combinations of `barangay` + `service_type` + `date_filed` could still narrow results down to a small number of residents in low-volume barangays, even without a name or phone number. Aggregate to barangay-level counts before any public reporting, and avoid publishing row-level records.
